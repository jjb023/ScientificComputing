## Question 1

### a

```python
import numpy as np

def leibniz_pi_approximation(N):
    pi_approx = 0
    for n in range(N+1):
        pi_approx += 8 / ((4*n + 1) * (4*n + 3))
    return pi_approx

N_values = [100, 1000, 10000]

for N in N_values:
    pi_approx = leibniz_pi_approximation(N)
    print(f"Approximation of π for N = {N}: {pi_approx}")
```

### b

```python
from math import pi

true_pi = pi

for N in N_values:
    pi_approx = leibniz_pi_approximation(N)
    error = abs(true_pi - pi_approx)
    print(f"Error for N = {N}: {error}")
    
```

### c

```python
N = 10000
while True:
    N += 1
    pi_approx = leibniz_pi_approximation(N)
    error = abs(true_pi - pi_approx)
    if error < 1e-7:
        break
print(f"N needed to produce error less than 1e-7: {N}")
```

## Question 3

### a

```python
A = np.array([[1, 0, 0, 0],
              [1, -2, 1, 0],
              [0, 1, -2, 1],
              [0, 0, 0, 1]])

b = np.array([0, 1, 1, 2])

x = np.linalg.solve(A, b)

print("Solution x:")
print(x)

Ax = np.dot(A, x)

difference = Ax - b

print("\nAx - b:")
print(difference)
```

When you run this code, it will print the solution x of the linear system, followed by the result of Ax−b.

The values of Ax−b should ideally be all zeros if the solution is exact. Any non-zero values indicate the discrepancy between the left-hand side and the right-hand side of the equation, meaning that the solution might not be exact or there could be numerical errors. In many cases, due to numerical precision, you might observe very small non-zero values instead of exactly zero.

### b

```python
# Plot y as a function of x. Add labels to the x and y axes. Edit the line colour and
# thickness and font sizes to your preference.

import matplotlib.pyplot as plt

t = np.linspace(0, 5, 500)
y = t**2 * np.exp(-2*t)

plt.plot(t, y, color='red', linewidth=2)
plt.xlabel('t')
plt.ylabel('y')
plt.show()
```

### c
