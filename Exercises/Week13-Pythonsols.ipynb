## Question 1

### a

```python
import numpy as np

def leibniz_pi_approximation(N):
    pi_approx = 0
    for n in range(N+1):
        pi_approx += 8 / ((4*n + 1) * (4*n + 3))
    return pi_approx

# Compute approximations for N = 100, 1000, and 10000
N_values = [100, 1000, 10000]

for N in N_values:
    pi_approx = leibniz_pi_approximation(N)
    print(f"Approximation of π for N = {N}: {pi_approx}")
```

### b

```python
from math import pi

true_pi = pi

# Compute the relative error for each approximation
for N in N_values:
    pi_approx = leibniz_pi_approximation(N)
    error = abs(true_pi - pi_approx)
    print(f"Error for N = {N}: {error}")
    
```

### c

```python
# Compute value of N needed to produce error less than 1e-7
N = 0
while True:
    N += 100000
    pi_approx = leibniz_pi_approximation(N)
    error = abs(true_pi - pi_approx)
    if error < 1e-7:
        break
print(f"N needed to produce error less than 1e-7: {N}")
```
